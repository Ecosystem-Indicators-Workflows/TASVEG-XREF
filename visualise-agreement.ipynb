# Visualise disagreements

We want to calculate measures of overall agreement or disagreement, and pairwise agreement measures (differences between experts). 

Patterns of disagreement: frequency of agreement of all experts, three experts or only two experts, etc.

We can look at the problem from three angles:

- for each *input type*, how many expert assign the same group/membership? what is the disagreement between experts?
- for each *experts*, how many types do they assign to each group

## Summarise the data

Now that we have the input from four experts in the same format, we can produce some informative summaries:

```r
(candidates_per_vegcode <- xref_table_experts |>
  group_by(VEGCODE) |>
  summarise(`Nr. of candidates`=n_distinct(efg), `Highest membership`=max(memb)))
```

For example, how many candidate functional groups do we have per type (identified by `VEGCODE`) 

```r
candidates_per_vegcode |>
  pull(`Nr. of candidates`) |>
  table()
```

```r
xref_table_experts |>
  group_by(VEGCODE, efg, memb) |>
  summarise(`Nr. of experts`=n_distinct(expert), 
    experts=paste(expert,collapse=";"),
    .groups="drop") |>
  arrange(VEGCODE, desc(`Nr. of experts`), desc(memb))
```

```r
xref_table_experts |>
  group_by(efg, expert) |>
  summarise(`types`=n_distinct(VEGCODE), 
    `membership total`=sum(memb),
    .groups="drop") 
```

Create the heatmap plot


```r
#| fig-width: 3
#| fig-height: 9
#| out-height: "170%"
#| fig-align: "center"
#| label: fig-polar
#| echo: false
#| fig-cap: "Heatmap plot to check differences between experts for a subset of Ecosystem Functional Groups"
xref_table_experts |>
  group_by(efg, expert) |>
  summarise(`types`=n_distinct(VEGCODE), 
    `membership total`=sum(memb, na.rm=TRUE),
    .groups="drop") |>
  filter(grepl("^T",efg)) |>
ggplot(aes(y = efg, x = expert, fill = `membership total`)) +
  geom_tile(color = "white") +                    # Draws the grid cells
  geom_text(aes(label = types), color = "black",size=2) +    # Adds frequency numbers inside cells
  scale_fill_gradient(low = "lightblue", high = "darkblue") + # Color scale
  theme_minimal()
```

This plot can reveal errors in nomenclature ("TF7.2", presumably refering to "T7.2") or genuine differences between experts (only one expert assigning "TF1.5" or "TF1.4")

```r
mtz <- xref_table_experts |>
  group_by(efg, expert) |>
  summarise(`membership`=sum(memb, na.rm=TRUE),
    .groups="drop") |>
  pivot_wider(names_from = "efg", values_from="membership")

#vegdist(mtz)
```

```r
dissim_matrix <- dist(mtz[,-1], method = "euclidean")
```

```r
plot(hclust(dissim_matrix,),
 labels=mtz$expert ,
     main = "Dissimilarity Between Experts", 
     xlab = "Experts", 
     ylab = "Distance")
```

## Pairwise diferences in membership

```r
mtz <- pivot_wider(
  xref_table_experts,
  names_from = "expert", values_from="memb") %>%
    replace(is.na(.),0)
mtz
```

```r

  mtz |>
    filter(AK>0 | JB>0) |>
    summarise(mean(AK-JB))
```

```r
mtz |>
    filter(AK>0 | MGT>0) |>
    mutate(disagreement=abs(AK-MGT)) |>
    group_by(efg) |>
    summarise(
      n=n_distinct(VEGCODE),
      disagreement=mean(disagreement)) |>
    arrange(desc(disagreement))
```
